functools

functools is a standard Python module that provides tools for working with functions and callable objects.

It is especially useful for:

modifying functions

creating reusable functions

caching function results

reducing collections of values

creating partial functions

comparing objects

building decorators

```python
from functools import reduce
```

Important functools functions

reduce()
Combine multiple values into one

partial()
Create a new function with some arguments already fixed

partialmethod()
Same idea as partial(), mainly for class methods

lru_cache()
Cache function results

cache()
Simple unlimited caching

cached_property()
Cache a class property

wraps()
Preserve metadata when writing decorators

update_wrapper()
Copy metadata from one function to another

cmp_to_key()
Convert old-style comparison functions for sorting

total_ordering()
Generate comparison methods for a class

singledispatch()
Function overloading based on the first argument's type

singledispatchmethod()
Same concept for methods

Placeholder
Reserve positional argument positions in partial()

reduce() repeatedly applies a function to the elements of an iterable and produces one final value.

syntax :- reduce(function, iterable)

reduce(function, iterable, initializer)

```python
numbers = [1, 2, 3, 4]
result = reduce(lambda x, y: x + y, numbers)
print(result)
```

partial() creates a new function with some arguments of an existing function already filled in.

```python
def power(base, exponent):
    return base ** exponent
print(power(2, 3))
print(power(5, 3))
```

```python
from functools import partial,partialmethod
square = partial(power, exponent=2)
print(square(5))
print(square(10))
```

```python
def multiply(a, b):
    return a * b
mult_by_10 = partial(multiply,10)
print(mult_by_10(5))
print(mult_by_10(9))
```

```python
multiply_by_10 = lambda x: multiply(10, x)
multiply_by_10(10)
```

partialmethod() is similar to partial(), but it is designed for methods inside classes.

```python
class Calculator:
    def multiply(self, x, y):
        return x * y
    double = partialmethod(multiply, 2)

c = Calculator()
print(c.double(10))
```

lru_cache() stores previously calculated results so that if the function is called again with the same arguments, Python can reuse the stored result.

LRU means: Least Recently Used

```python
from functools import lru_cache
@lru_cache
def square(n):
    print("Calculating...")
    return n * n

print(square(5))
print(square(5))
```

```python
def fib(n):
    if n <= 1:
        return n
    return fib(n - 1) + fib(n - 2)

fib(8)
```

```python
@lru_cache
def fib(n):
    if n <= 1:
        return n
    return fib(n - 1) + fib(n - 2)
print(fib(7))
fib.cache_info()
```

```python
fib.cache_clear()
```

cache() is a simpler form of memoization for functions whose arguments are hashable.

```python
from functools import cache
@cache
def square(n):
    return n * n
square(5)
```

```python
from functools import cache
@cache
def square(n):
    print("Calculating...")
    return n * n

print(square(5))
print(square(5))
print(square(5))
print(square(10))
```

cache() is essentially an unbounded cache.

lru_cache() allows you to control the maximum number of cached entries.

cached_property() turns a method into a property whose result is calculated once and then stored on the instance.

```python
from functools import cached_property
class Student:
    def __init__(self, marks):
        self.marks = marks
    @cached_property
    def percentage(self):
        print("Calculating...")
        return sum(self.marks) / len(self.marks)

s = Student([80, 90, 70])
print(s.percentage)
```

update_wrapper() is the function underlying the idea used by wraps().

update_wrapper() copies important metadata from one function to another function.

```python
from functools import update_wrapper
def original():
    """Original function"""
    pass
def wrapper():
    pass
update_wrapper(wrapper, original)
```

```python
from functools import update_wrapper
def my_decorator(func):
    def wrapper():
        print("Before")
        func()
    update_wrapper(wrapper, func)
    return wrapper
@my_decorator
def hello():
    """This is hello function."""
    print("Hello")
print(hello.__name__)
print(hello.__doc__)
```

```python
from functools import update_wrapper
def my_decorator(func):
    def wrapper():
        print("Before function")
        func()
        print("After function")
    update_wrapper(wrapper, func)
    return wrapper
@my_decorator
def hello():
    """This is the hello function."""
    print("Hello")

hello()
print(hello.__name__)
print(hello.__doc__)
```

cmp_to_key() converts a comparison function into a key function that sorting functions can use.

```python
from functools import cmp_to_key
def compare(a, b):
    if a > b:
        return 1
    elif a < b:
        return -1
    else:
        return 0

numbers = [5, 2, 8, 1]
result = sorted(numbers, key=cmp_to_key(compare))

print(result)
```

total_ordering() can generate several ordering methods when you provide the required comparison information.

```python
from functools import total_ordering
@total_ordering
class Student:
    def __init__(self, marks):
        self.marks = marks
    def __eq__(self, other):
        return self.marks == other.marks
    def __lt__(self, other):
        return self.marks < other.marks
```

singledispatch() allows different implementations of a function depending on the type of its first argument.

```python
from functools import singledispatch

@singledispatch
def show(value):
    print("Unknown type")
```

```python
@show.register
def _(value: int):
    print("Integer:", value)
```

```python
@show.register
def _(value: str):
    print("String:", value)
```

```python
show(10)
show("Python")
```

```python
def subtract(a, b, c):
    return a - b - c
print(subtract(10, 3, 2))
```

```python
from functools import partial
subtract_3 = partial(subtract, b=3)

print(subtract_3(10,c=2))
```

```python
from functools import partial, Placeholder as _

def subtract(a, b, c):
    return a - b - c

new_function = partial(subtract, _, 10, 2)

print(new_function(20))
```

```python
from functools import partial, Placeholder as _

def calculate(a, b, c):
    return a + b * c

f = partial(calculate, _, 10, 2)

print(f(5))
```

```python
def power(a, b):
    return a ** b
square = partial(power, _, 2)
print(square(5))
print(square(10))
```

```python
#singledispatchmethod()
from functools import singledispatchmethod
class Printer:

    @singledispatchmethod
    def print_value(self, value):
        print("Unknown:", value)

    @print_value.register
    def _(self, value: int):
        print("Integer:", value)

    @print_value.register
    def _(self, value: str):
        print("String:", value)

    @print_value.register
    def _(self, value: float):
        print("Float:", value)

    @print_value.register
    def _(self, value: list):
        print("List:", value)


p = Printer()

p.print_value(10)
p.print_value("Hello")
p.print_value(10.5)
p.print_value([1, 2, 3])
p.print_value(True)
```
