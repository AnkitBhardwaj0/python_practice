1. timestamp

January 1, 1970, 00:00:00 UTC

UTC = Coordinated Universal Time.

It is the global reference time used to define time zones.

5. time.localtime()

Converts a timestamp into local time components.

struct_time object containing things such as:

year
month
day
hour
minute
second
weekday
day of year
DST flag

DST = Daylight Saving Time.

Some countries/regions change their clocks during part of the year, usually by moving them one hour forward.

| Value | Meaning                                                  |
| ----: | -------------------------------------------------------- |
|   `0` | DST is not active                                        |
|   `1` | DST is active                                            |
|  `-1` | Information is unavailable / let the system determine it |


```python
t = time.localtime()
print(t)
```

```python
#Indicates whether the system timezone has a DST rule
time.daylight
```

```python
t = time.localtime()
print(t.tm_year)
print(t.tm_mon)
print(t.tm_mday)
print(t.tm_hour)
print(t.tm_min)
print(t.tm_sec)
```

6. time.gmtime()

gmtime() is similar to localtime(), but it gives UTC time.

```python
t = time.gmtime()
print(t)
```

localtime() vs gmtime()

| Function           | Time       |
| ------------------ | ---------- |
| `time.localtime()` | Local time |
| `time.gmtime()`    | UTC        |

IST = UTC + 5:30  local time will be 5 hours 30 minutes ahead of UTC.


7. time.strftime()
This is extremely important for formatting dates and times.
strftime() means roughly:

Important format codes
%Y - Four-digit year:2026

%y - Two-digit year:26

%B - month name :August

%b - short month name :aug

%m - Month:08

%A - day name :wednesday

%a - short name :wed

%d - Day:19

%H - Hour in 24-hour format:22

%I - Hour in 12-hour format:10

%M - Minute:20

%S - Second:15

%p - AM/PM:PM

```python
t = time.localtime()
formatted = time.strftime("%d-%m-%Y", t)
print(formatted)
```

```python
t = time.localtime()
print(time.strftime("%d-%m-%Y", t))
print(time.strftime("%H:%M:%S", t))
print(time.strftime("%d/%m/%Y %I:%M:%S %p", t))
```

8. time.strptime()

This does almost the opposite of strftime().
strptime() means:
string parsed into time


```python
date = "19-08-2026"
t = time.strptime(date, "%d-%m-%Y")
print(t)
```

strftime() vs strptime()
strftime
    ↓
time → string

strptime
    ↓
string → time structure

strftime = formatting
strptime = parsing

10. time.perf_counter()

This is especially useful for measuring how long code takes to execute.

Why use perf_counter()?
For benchmarking code, perf_counter() is generally preferred over time.time() because it is designed for measuring elapsed durations with high-resolution timing.

3.229999856557697e-05 :- 3.229999856557697 × 10⁻⁵ seconds :- 0.0000323 seconds :-0.0323 milliseconds :-32.3 microseconds

```python
start = time.perf_counter()
for i in range(1000000):
    pass
end = time.perf_counter()
print(end - start)
```

```python
start = time.perf_counter()
# code
end = time.perf_counter()
elapsed = end - start
print(elapsed)
```

11. time.process_time()

Measures CPU time consumed by the current process.
process_time() does not count time spent sleeping.

```python
start = time.process_time()
for i in range(1000000):
    pass
end = time.process_time()
print(end - start)
```

```python
start = time.process_time()
time.sleep(2)
for i in range(1000000):
    pass
end = time.process_time()
print(end - start)
```

Important timing functions

| Function              | Main purpose                    |
| --------------------- | ------------------------------- |
| `time.time()`         | Current Unix timestamp          |
| `time.sleep()`        | Pause execution                 |
| `time.localtime()`    | Local `struct_time`             |
| `time.gmtime()`       | UTC `struct_time`               |
| `time.ctime()`        | Readable time string            |
| `time.strftime()`     | Time → formatted string         |
| `time.strptime()`     | String → `struct_time`          |
| `time.mktime()`       | Local `struct_time` → timestamp |
| `time.perf_counter()` | Measure elapsed time            |
| `time.process_time()` | Measure CPU time                |

Login Attempt Timer

Create a program where the user has 3 attempts to enter a password.
for example:

Enter password: hello
Wrong password

Enter password: python
Wrong password

Enter password: admin123
Login successful 

Measure the total time taken by the user to successfully log in.


```python
password='admin123'
start=time.perf_counter()
for i in range(3):
     enter_password=input('enter password')
     if password==enter_password:
          print('login successful')
          break
     else:
          print('unsuccessful')
end=time.perf_counter()
print(f'time taken in login : {end-start} ')
```

Code Performance Comparison

Compare the execution time of these two approaches:

```python
# Method 1
start_1=time.perf_counter()
result = []
for i in range(1000000):
    result.append(i * 2)
end_1=time.perf_counter()
time_taken_1=end_1-start_1

# Method 2
start_2=time.perf_counter()
result = [i * 2 for i in range(1000000)]
end_2=time.perf_counter()
time_taken_2=end_2-start_2

print(time_taken_1,'\n',time_taken_2)
diff=abs(time_taken_1-time_taken_2)
print(f'diff : {diff} ')
```
