reduce :- reduce() is a function from Python's functools module.

It is used to apply a function repeatedly to the elements of an iterable and reduce them to a single final value.

reduce() takes multiple values and combines them step-by-step to produce one value.

Syntax

reduce(function, iterable)

reduce(function, iterable, initializer)

    function,      ← operation to perform

    iterable,      ← data
    
    initializer    ← optional starting value

```python
from functools import reduce
```

reduce() combines two values at a time.

x = accumulated result

y = next element

```python
numbers = [1, 2, 3, 4, 5]
result = reduce(lambda x,y: x + y, numbers)
print(result)
```

Compare this with map():

map() normally processes one element at a time, so the lambda has one parameter:

```python
def add(x, y):
    return x + y
numbers = [1, 2, 3, 4, 5]
result = reduce(add, numbers)
print(result)
```

```python
numbers = [1, 2, 3, 4, 5]
result = reduce(lambda x, y: x * y, numbers)
print(result)
```

reduce() can also be used to find the largest value.

```python
numbers = [10, 45, 23, 89, 34, 12]
result = reduce(lambda x, y: x if x > y else y, numbers)
print(result)
```

```python
words = ["Python", "is", "very", "powerful"]
result = reduce(lambda x, y: x + " " + y, words)
print(result)
```
