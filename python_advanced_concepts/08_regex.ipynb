Match Object

search(), match(), fullmatch() etc. successful match par Match object return kar sakte hain.

```python
""" 
.group() : Matched text:
.start() : Match ka starting index. 
.end() : Match ke ending index se one past the last matched character.
.span() : Dono:
"""
text = "My age is 21"
m = re.search(r"\d+", text)
print(m.group())
print(m.start())
print(m.end())
print(m.span())
```

Grouping ( ) : Parentheses pattern ko group karne ke liye

```python
text = "Date: 07-09-2026"
m=re.search(r"(\d{2})-(\d{2})-(\d{4})",text)
print(m.group())
print(m.group(1))
print(m.group(2))
print(m.group(3))
```

named groups : groups ko name bhi de sakte hai 

synatx :(?P<name>pattern)

Ye large Regex patterns me readability improve karta hai.

```python
text = "Name: Ankit, Age: 21"
pattern = r"Name: (?P<name>\w+), Age: (?P<age>\d+)"
m = re.search(pattern, text)

print(m.group("name"))
print(m.group("age"))
```

```python
m.group()
```

Regex Flags

Regex behavior modify karne ke liye flags use hote hain.

re.IGNORECASE (re.I): Upper/lowercase difference ignore karta hai.

re.MULTILINE : ^ and $ ko multiple lines ke context me work karne deta hai.

re.DOTALL : Normally . newline ko match nahi karta.DOTALL lagane par . newline ko bhi match kar sakta hai.

re.ASCII (re.A) : regex behavior ko ASCII characters tak restrict karta hai for certain shorthand character classes.

```python
text = "Python PYTHON python"
print(re.findall(r"python", text, re.IGNORECASE))
```

```python
text = "Python PYTHON python"
print(re.findall(r"python", text, re.I))
```

```python
text = """Python
Java
Python
C"""
print(re.findall(r"^Python", text, re.MULTILINE))
```

```python
text = "Hello\nPython"
result = re.search(r"Hello.Python", text, re.DOTALL)
print(result.group())
```

```python
re.findall(r"\w+", text, re.A)
```

Greedy vs Non-Greedy

Quantifiers by default greedy hote hain.

```python
text = "<h1>Hello</h1><h1>World</h1>"
result = re.findall(r"<h1>.*</h1>", text)
print(result)
```

Greedy .* maximum possible text consume kar sakta hai.

Non-greedy version:

 r"<h1>.*?</h1>" : ? yahan quantifier ko non-greedy bana raha hai.

```python
result = re.findall(r"<h1>.*?</h1>", text)
print(result)
```

Lookahead 

Positive lookahead : 

r"\d+(?= dollars)" : Number tab match hoga jab uske baad " dollars" ho.

syntax : (?=pattern)

Negative lookahead :

(?!pattern) :- Means next position par pattern nahi hona chahiye.

Lookbehind

Positive lookbehind

(?<=pattern)

Negative lookbehind 

(?<!pattern) : Means given pattern immediately before nahi hona chahiye.

```python
text = "₹500 $200"
print(re.findall(r"(?<=₹)\d+", text)) #Because number ke immediately pehle ₹ hona chahiye.
```

```python
email="ankit@gmail.com"
pattern=r"^[A-Za-z0-9%_.+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$"
result=re.fullmatch(pattern,email)
if result:
     print("valid")
else:
     print("invalid")
```

```python
#10 digit phone validation
number="9876564543"
pattern=r"[6-9]\d{9}"
if re.fullmatch(pattern,number):
     print("valid number")
else:
     print("invalid number")
```

```python
name="Ankit"
if re.fullmatch(r"[A-Za-z]+",name):
     print("valid name")
else:
     print("invalid")
```

```python
number = "123456"
if re.fullmatch(r"\d+", number):
    print("Valid")
```

```python
#Username validation
""" 
letters allowed
digits allowed
underscore allowed
length 3-15
"""
pattern=r"[A-Za-z0-9_]{3,15}"
if re.fullmatch(pattern,"ankit12"):
     print("valid")
else:
     print("invalid")
```

Password validation

Example requirement:

minimum 8 characters
at least one uppercase
at least one lowercase
at least one digit
at least one special character

```python
password="aAnkit12@"
pattern=(r"^(?=.*[A-Z])"r"(?=.*[a-z])"r"(?=.*\d)"r"(?=.*[@$&!*?%])"r".{8,}$")
if re.fullmatch(pattern, password):
    print("Strong password")
else:
    print("Weak password")
```
