timedelta

timedelta represents a duration.

```python
from datetime import timedelta
duration = timedelta(days=5)
print(duration)
```

```python
timedelta(days=5,hours=3,minutes=20,seconds=10)
```

```python
today = date.today()
future = today + timedelta(days=10)
print(today)
print(future)
```

```python
today = date.today()
past = today - timedelta(days=10)
print(past)
```

```python
now = datetime.now()
future = now + timedelta(hours=5)
print(future)
```

```python
timedelta(weeks=1,days=2,hours=5,minutes=30,seconds=20)
```

Important format codes

| Code | Meaning             | Example    |
| ---- | ------------------- | ---------- |
| `%Y` | 4-digit year        | `2026`     |
| `%y` | 2-digit year        | `26`       |
| `%m` | month               | `08`       |
| `%d` | day                 | `20`       |
| `%H` | 24-hour             | `19`       |
| `%I` | 12-hour             | `07`       |
| `%M` | minute              | `45`       |
| `%S` | second              | `30`       |
| `%f` | microsecond         | `123456`   |
| `%p` | AM/PM               | `PM`       |
| `%A` | full weekday        | `Thursday` |
| `%a` | abbreviated weekday | `Thu`      |
| `%B` | full month          | `August`   |
| `%b` | abbreviated month   | `Aug`      |


```python
date_string = "20-08-2026"
dt = datetime.strptime(
    date_string,
    "%d-%m-%Y"
)
print(dt)
```

```python
date_time_string = "20-08-2026 19:30:45"
dt = datetime.strptime(
    date_time_string,
    "%d-%m-%Y %H:%M:%S"
)
print(dt)
```

datetime.timestamp()

A datetime can be converted into a Unix timestamp.

```python
dt = datetime.now()
timestamp = dt.timestamp()
print(timestamp)
```

```python
timestamp = 1787145000
dt = datetime.fromtimestamp(timestamp)
print(dt)
```

```python
timestamp = 1787145000
dt = datetime.fromtimestamp(timestamp,timezone.utc)
print(dt)
```

```python
dt = datetime.now()
print(dt.tzinfo)
```

```python
dt = datetime.now(timezone.utc)
print(dt.tzinfo)
```

```python
from zoneinfo import ZoneInfo
india_time = datetime.now(ZoneInfo("Asia/Kolkata"))
print(india_time)
```

```python
new_york_time = datetime.now(ZoneInfo("America/New_York"))
print(new_york_time)
```

```python
india = datetime.now(ZoneInfo("Asia/Kolkata"))
new_york = india.astimezone(ZoneInfo("America/New_York"))
print(india)
print(new_york)
```

```python
birth_date = date(2000, 5, 10)
today = date.today()
age = today.year - birth_date.year
if (today.month, today.day) < (birth_date.month, birth_date.day):
    age -= 1
print(age)
```
