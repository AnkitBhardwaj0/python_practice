filter()

filter() is a built-in Python function used to select elements from an iterable based on a condition.

filter() keeps only the elements for which the condition is True.

Syntax

filter(function, iterable)

```python
def check_even(x):
    return x % 2 == 0
numbers = [1, 2, 3, 4, 5, 6]
result = filter(check_even, numbers)
print(list(result))
```

```python
numbers = [1, 2, 3, 4, 5, 6]
odd_number = filter(lambda x: x % 2 != 0, numbers)
print(list(odd_number))
```

```python
numbers = [1, 2, 3, 4, 5, 6]
even_numer=list(filter(lambda x : x%2==0,numbers))
print(even_numer)
```

```python
even_number = filter(lambda x: x % 2 == 0, numbers)
print(even_number)
```

```python
numbers = [5, 12, 8, 20, 3, 15]
result = list(filter(lambda x: x > 10, numbers))
print(result)
```

```python
#Positive Numbers
numbers = [-5, 10, -2, 8, -1, 20]
result = list(filter(lambda x: x > 0, numbers))
print(result)
```

```python
#Negative Numbers
numbers = [-5, 10, -2, 8, -1, 20]
result = list(filter(lambda x: x < 0, numbers))
print(result)
```

```python
#Divisible by 5
numbers = [10, 12, 15, 20, 23, 25, 30]
result = list(filter(lambda x: x % 5 == 0, numbers))
print(result)
```

Filtering Strings



```python
names = ["Ankit", "Aman", "Rahul", "Ajay", "Rohit"]
result = list(filter(lambda name: name.startswith("A"), names))
print(result)
```

```python
result = list(filter(lambda name: len(name) > 4, names))
print(result)
```

```python
result = list(filter(lambda name: len(name) % 2 == 0, names))
print(result)
```

Filtering a List of Dictionaries



```python
students = [{"name": "Ankit", "marks": 80},{"name": "Rahul", "marks": 45},{"name": "Aman", "marks": 70},{"name": "Rohit", "marks": 35}]
result = list(filter(lambda student: student["marks"] >= 50, students))
print(result)
```

Multiple Conditions

```python
numbers = [10, 15, 20, 25, 30, 35, 40]
result = list( filter(lambda x: x > 20 and x < 40, numbers))
print(result)
```

```python
result = list(filter(lambda x: x < 15 or x > 35, numbers))
print(result)
```

```python
def check(x):
    return x > 10
numbers = [5, 12, 8, 20, 15]
result = list(filter(check,numbers))
print(result)
```

filter() Returns an Iterator

```python
numbers = [1, 2, 3, 4]
result = filter(lambda x: x % 2 == 0, numbers)
print(type(result))
for x in result:
    print(x)
```

Filter Object is Consumed

```python
numbers = [1, 2, 3, 4]
result = filter(lambda x: x % 2 == 0, numbers)
print(list(result))
print(list(result))
```

filter(None, iterable)

When the function is None, filter() keeps elements whose Boolean value is True.

```python
#filter(None, iterable)
numbers = [0, 1, False, True, "", "Hello", None, 5]
result = list(filter(None,numbers))
print(result)
```

filter() with Tuples

```python
numbers = (10, 15, 20, 25, 30)
result = tuple(filter(lambda x: x % 10 == 0, numbers))
print(result)
```

filter() with Sets

```python
numbers = {10, 15, 20, 25, 30}
result = list(filter(lambda x: x > 20, numbers))
print(result)
```

```python
numbers = list(map(int, input().split()))
result = list(filter(lambda x: x % 5 == 0, numbers))
print(result)
```

```python
numbers=[1,2,3,4,5,6]
result = list(map(lambda x: x * 10,filter(lambda x: x % 2 == 0, numbers)))
print(result)
```
