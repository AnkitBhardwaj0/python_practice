```python
cols=['District code','District name', 'Population','Male','Female','Literate','Male_Literate','Female_Literate','Housholds_with_Electric_Lighting','Households_with_Internet']
```

```python
census=census[cols]
census.head()
```

```python
final_df=latlong.merge(census,left_on='District',right_on='District name').drop(columns='District name')
```

```python
final_df
```

```python
final_df['sex_ratio']=round((final_df['Female']/final_df['Male'])*100)
```

```python
final_df['litracy_rate']=round((final_df['Literate']/final_df['Population'])*100)
```

```python
final_df.drop(columns=['Male','Female','Literate'],inplace=True)
```

```python
final_df
```

```python
final_df.to_csv('india.csv',index=False)
```
