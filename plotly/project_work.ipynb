```python
cols=['District code','District name', 'Population','Male','Female','Literate','Male_Literate','Female_Literate','Housholds_with_Electric_Lighting','Households_with_Internet']
```

```python
census=census[cols]
census.head()
```

```python
final_df=latlong.merge(census,left_on='District',right_on='District name').drop(columns='District name')
```

```python
final_df
```

```python
final_df['sex_ratio']=round((final_df['Female']/final_df['Male'])*100)
```

```python
final_df['litracy_rate']=round((final_df['Literate']/final_df['Population'])*100)
```

```python
final_df.drop(columns=['Male','Female','Literate'],inplace=True)
```

```python
final_df
```

```python
final_df.to_csv('india.csv',index=False)
```

```python
import plotly.express as px
df = px.data.carshare()
fig = px.scatter_map(df, lat="centroid_lat", lon="centroid_lon", color="peak_hour", zoom=10,map_style="open-street-map")
fig.show()   
```

```python
df[['centroid_lat', 'centroid_lon', 'peak_hour']].head()
```

```python
print(df["centroid_lat"].min(), df["centroid_lat"].max())
print(df["centroid_lon"].min(), df["centroid_lon"].max())
```

```python
center = {
    "lat": df["centroid_lat"].mean(),
    "lon": df["centroid_lon"].mean()
}

fig = px.scatter_map(
    df,
    lat="centroid_lat",
    lon="centroid_lon",
    color="peak_hour",
    center=center,
    zoom=10,
    map_style="open-street-map"
)

fig.show()
```
