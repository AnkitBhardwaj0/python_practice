```python
import numpy as np
import pandas as pd
import plotly.graph_objects as go
import plotly.offline as pyo
```

```python
matches=pd.read_csv('matches.csv')
delivery=pd.read_csv('deliveries.csv')
ipl=delivery.merge(matches,left_on='match_id',right_on='id')
ipl.head()
```

```python
#scatter plots
#problem :- we are going to draw a scatter plot between batsman avg (x-axis) and 
# batsman strike rate on y asix of top 50 batsman in ipl all time
```

```python
#avg vs sr graph of top 50 batsman (in term of total runs)

top50=ipl.groupby('batsman')['batsman_runs'].sum().sort_values(ascending=False).head(50).index.tolist()
new_ipl=ipl[ipl['batsman'].isin(top50)]
```

```python
#calculating sr
runs=new_ipl.groupby('batsman')['batsman_runs'].sum()
balls=new_ipl.groupby('batsman')['batsman_runs'].count()
sr=(runs/balls)*100
sr=sr.reset_index()
sr
```

```python
#calculating avg
#avg=(total number of runs)/(number of outs)
out=ipl[ipl['player_dismissed'].isin (top50)]
nout=out['player_dismissed'].value_counts()
avg=runs/nout
avg=avg.reset_index()
avg
```

```python
df=avg.merge(sr,on='batsman')
df.rename(columns={0:'avg','batsman_runs':'sr'},inplace=True)
df
```

```python
#scatter plot

trace=go.Scatter(x=df['avg'],y=df['sr'],mode='markers',hovertext=df['batsman'],marker={'color':'#00a65a'})
data=[trace]
layout=go.Layout(title='batsman Avg vs sr',xaxis={'title':'Avg'},yaxis={'title':'sr'})
fig=go.Figure(data=data,layout=layout)
#pyo.plot(fig) #show in server
fig.show()
```

```python
trace=go.Scatter(x=df['avg'],y=df['sr'],mode='markers',hovertext=df['batsman'],marker={'color':'#00a65a','size':16})
data=[trace]
layout=go.Layout(title='batsman Avg vs sr',xaxis={'title':'Avg'},yaxis={'title':'sr'})
fig=go.Figure(data=data,layout=layout)
fig.show()
```

```python
#multiple line charts
def batsman_comp(*name):
     data=[]
     for i in name:
          single=ipl[ipl['batsman']==i]
          performance=single.groupby('season')['batsman_runs'].sum().reset_index()
          trace=go.Scatter(x=performance['season'],y=performance['batsman_runs'],mode='lines+markers',name=i)
          data.append(trace)
          layout=go.Layout(title='year by year performance',xaxis={'title':'season'},yaxis={'title':'runs'})
          fig=go.Figure(data=data,layout=layout)
          fig.show()
```

```python
batsman_comp('V Kohli','RG Sharma','DA Warner','MS Dhoni')
```

### Bar plot

```python
top10=ipl.groupby('batsman')['batsman_runs'].sum().sort_values(ascending=False).head(10).index.tolist()
top10_df=ipl[ipl['batsman'].isin(top10)]
```

```python
top10_score=top10_df.groupby('batsman')['batsman_runs'].sum().reset_index()
top10_score
```

```python
#plot bar graph
trace=go.Bar(x=top10_score['batsman'],y=top10_score['batsman_runs'],marker={'color':'orange'})
data=[trace]
layout=go.Layout(title='top 10 ipl batsman',xaxis={'title':'batsman'},yaxis={'title':'batsman_run'})
fig=go.Figure(data=data,layout=layout)
fig.show()
```

> there are 2 types of Bar Graph
1. nested Bar graph
2. stacked Bar graph
3. overlayed Bar graph

```python
iw=top10_df.groupby(['batsman','inning'])['batsman_runs'].sum().reset_index()
mask=iw['inning']==1
mask2=iw['inning']==2
one=iw[mask]
two=iw[mask2]

one.rename(columns={'batsman_runs':'1st innings'},inplace=True)
two.rename(columns={'batsman_runs':'2nd innings'},inplace=True)
final=one.merge(two,on='batsman')[['batsman','1st innings','2nd innings']]
final
```

```python
trace1=go.Bar(x=final['batsman'],y=final['1st innings'],name='1st innings',marker={'color':'#00a65a'})
trace2=go.Bar(x=final['batsman'],y=final['2nd innings'],name='2nd innings',marker={'color':'#a6a65a'})

data=[trace1,trace2]
layout=go.Layout(title='inning by scores',xaxis={'title':'batsman'},yaxis={'title':'runs'},barmode='overlay')
fig=go.Figure(data=data,layout=layout)
fig.show()
```

```python
trace1=go.Bar(x=final['batsman'],y=final['1st innings'],name='1st innings',marker={'color':'#00a65a'})
trace2=go.Bar(x=final['batsman'],y=final['2nd innings'],name='2nd innings',marker={'color':'#a6a65a'})

data=[trace1,trace2]
layout=go.Layout(title='inning by scores',xaxis={'title':'batsman'},yaxis={'title':'runs'},barmode='stack')
fig=go.Figure(data=data,layout=layout)
fig.show()
```

```python
trace1=go.Bar(x=final['batsman'],y=final['1st innings'],name='1st innings',marker={'color':'#00a65a'})
trace2=go.Bar(x=final['batsman'],y=final['2nd innings'],name='2nd innings',marker={'color':'#a6a65a'})

data=[trace1,trace2]
layout=go.Layout(title='inning by scores',xaxis={'title':'batsman'},yaxis={'title':'runs'})
fig=go.Figure(data=data,layout=layout)
fig.show()
```

##### Box plot

```python
match_agg=delivery.groupby(['match_id'])['total_runs'].sum().reset_index()
season_wise=match_agg.merge(matches,left_on='match_id',right_on='id')[['match_id','total_runs','season']]
season_wise
```

```python
trace=go.Box(x=season_wise['total_runs'],name='all seasons',marker={'color':'#00a65a'})
data=[trace]
layout=go.Layout(title='Total Score Analysis',xaxis={'title':'Total Score'})

fig=go.Figure(data=data,layout=layout)

fig.show()
```

```python
trace1=go.Box(x=season_wise[season_wise['season']==2017]['total_runs'],name='2017',marker={'color':'#00a65a'})
trace2=go.Box(x=season_wise[season_wise['season']==2008]['total_runs'],name='2008',marker={'color':"#a6001e"})
data=[trace1,trace2]
layout=go.Layout(title='Total Score Analysis',xaxis={'title':'Total Score'})

fig=go.Figure(data=data,layout=layout)

fig.show()
```
