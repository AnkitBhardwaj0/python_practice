NumPy

NumPy (Numerical Python) is an open-source Python library used for numerical computing. Its main object is the ndarray, a multidimensional array designed for efficient operations on numerical data.

NumPy performs the operation element by element.

This type of operation is called vectorization.

```python
import numpy as np
```

```python
numbers = np.array([10, 20, 30, 40])
numbers * 2
```

NumPy Array — ndarray

An ndarray is NumPy's fundamental data structure. It represents a multidimensional, generally homogeneous collection of elements.

```python
a = np.array([10, 20, 30, 40])
print(a)
print(type(a))
```

np.linalg.det()

Matrix Multiplication — @

```python
A = np.array([
    [1, 2],
    [3, 4]
])
B = np.array([
    [5, 6],
    [7, 8]
])
C = A @ B
print(C)
```

np.matmul()

Performs matrix multiplication.

```python
np.matmul(A, B)
```

np.dot()

Performs dot-product operations. Its behavior depends on the dimensions of the inputs.


```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(np.dot(a, b))
```

Matrix Determinant

The determinant is a scalar value associated with a square matrix.

```python
A = np.array([
    [1, 2],
    [3, 4]
])
print(np.linalg.det(A))
```

Matrix Inverse

np.linalg.inv() calculates the inverse of a square, nonsingular matrix.

```python
A = np.array([
    [1, 2],
    [3, 4]
])
print(np.linalg.inv(A))
```

Solving Linear Equations

np.linalg.solve() solves a system of linear equations.

```python
A = np.array([
    [2, 1],
    [1, 3]
])
b = np.array([5, 6])
x = np.linalg.solve(A, b)
print(x)
```

Saving NumPy Arrays

np.save()

Saves a single NumPy array in .npy format.

```python
a = np.array([10, 20, 30])
np.save("data.npy", a)
```

```python
#load it
a = np.load("data.npy")
print(a)
```

np.savez()

Saves multiple NumPy arrays in a single .npz file.

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
np.savez("data.npz", first=a, second=b)
```

```python
data = np.load("data.npz")
print(data["first"])
print(data["second"])
```
