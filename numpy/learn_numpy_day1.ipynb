array operations

```python
a1=np.arange(12).reshape(3,4)
a2=np.arange(12,24).reshape(3,4)
print(a1)
print("\n")
print(a2)
```

```python
#scalar operation
#arithmetic
print(a1*2,'\n')
print(a1**2,'\n')
print(a1/2,'\n')
print(a1//2,'\n')
print(a1%2,'\n')
```

```python
#relational
print(a2>5,'\n')
print(a2>15)
```

```python
#vector operation
print(a1*a2,'\n')
print(a1*a2,'\n')
print(a1/a2,'\n')
print(a1//a2,'\n')
```

```python
a1=np.random.random((3,3))
a1=np.round(a1*100)
a1
```

```python
#max/min/sum/prod
print(np.max(a1))#max value in given array
print(np.min(a1))#minvalue in given array
print(np.sum(a1))#sum of all element of array
print(np.prod(a1))#product of all element in array

#0:- col and 1->row
print(np.max(a1,axis=1))#max value of each row
print(np.max(a1,axis=0))#max value of each column
print(np.min(a1,axis=0))
print(np.sum(a1,axis=1))#row by sum of element
```

```python
#mean/median/std/var
#same rule
print(np.mean(a1))
print(np.median(a1))
print(np.std(a1))
print(np.var(a1))
print(np.mean(a1,axis=0))
```

```python
#trigometric function
np.sin(a1)
```

```python
#dot product or vector product
a2=np.arange(12).reshape(3,4)
a3=np.arange(12,24).reshape(4,3)
np.dot(a2,a3)#result=(3,3) matrix
```

```python
#log and exponent
print(np.log(a1))
print("\n")
print(np.exp(a1))
```

stacking

```python
#horizontal 
a4=np.arange(12).reshape(3,4)
a5=np.arange(12,24).reshape(3,4)
a5
```

```python
a4
```

```python
np.hstack((a4,a5))
```

```python
np.hstack((a4,a5,a4))#add many but shape are same
```

```python
#vartical
a4=np.arange(12).reshape(3,4)
a5=np.arange(12,24).reshape(3,4)
```

```python
np.vstack((a4,a5))
```

```python
np.vstack((a4,a5,a4))
```
