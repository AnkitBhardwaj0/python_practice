### Colored Scatterplots

```python
iris=pd.read_csv('iris.csv')
iris.sample(5)
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'])
plt.xlabel('sepal length')
plt.ylabel('petal length')
```

```python
iris['Species']=iris['Species'].replace({'Iris-setosa': 0,'Iris-versicolor': 1,'Iris-virginica': 2})
```

```python
iris.sample(5)
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'])
plt.xlabel('sepal length')
plt.ylabel('petal length')
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='winter')
plt.xlabel('sepal length')
plt.ylabel('petal length')
#cmap : color variation change karta 
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='winter')
plt.xlabel('sepal length')
plt.ylabel('petal length')
plt.colorbar()
```

```python
# cmap and alpha
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='winter',alpha=0.5)
plt.xlabel('sepal length')
plt.ylabel('petal length')
plt.colorbar()
#alpha : color kaisa dikhe(0 to 1)ke bich
```

### 3D Surface Plots

```python
x=np.linspace(-10,10,100)
y=np.linspace(-10,10,100)
```

```python
xx,yy=np.meshgrid(x,y)
xx.shape #same y
```

```python
zz=xx**2+yy**2
zz.shape
```

```python
fig=plt.figure(figsize=(12,8))
ax=plt.subplot(projection='3d')
p=ax.plot_surface(xx,yy,zz,cmap='viridis')
fig.colorbar(p)
plt.show()
```

```python
zz=np.sin(xx)+np.cos(yy)
fig=plt.figure(figsize=(12,8))
ax=plt.subplot(projection='3d')
p=ax.plot_surface(xx,yy,zz,cmap='viridis')
fig.colorbar(p)
plt.show()
```

```python
zz=np.sin(xx)
fig=plt.figure(figsize=(12,8))
ax=plt.subplot(projection='3d')
p=ax.plot_surface(xx,yy,zz,cmap='viridis')
fig.colorbar(p)
plt.show()
```

```python
zz=xx**2+yy**2
fig=plt.figure(figsize=(12,8))
ax=plt.subplot(projection='3d')
p=ax.plot_surface(xx,yy,zz,cmap='viridis')
fig.colorbar(p)
plt.show()
```

### Contour Plots

```python
zz=xx**2+yy**2
fig=plt.figure(figsize=(12,8))
ax=plt.subplot()
p=ax.contour(xx,yy,zz,cmap='viridis')
fig.colorbar(p)
plt.show()
```

```python
zz=xx**2+yy**2
fig=plt.figure(figsize=(12,8))
ax=plt.subplot()
p=ax.contourf(xx,yy,zz,cmap='viridis')
fig.colorbar(p)
plt.show()
```

```python
temp_df=delivery[(delivery['ballnumber'].isin([1,2,3,4,5,6])) & (delivery['batsman_run']==6)]
```
