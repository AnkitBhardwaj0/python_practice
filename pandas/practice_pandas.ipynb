Pandas groupby() 

Basic syntax

df.groupby(by, axis=0, level=None, as_index=True, sort=True,group_keys=True, observed=True, dropna=True)

```python
import pandas as pd
import numpy as np
```

```python
df=pd.DataFrame({
     'Department':['IT','HR','IT','HR','Sales'],
     'salary':[50000,40000,60000,45000,55000]
})
df
```

```python
group=df.groupby('Department')
type(group)
```

Why doesn't groupby() itself give the result?
:-groupby() only performs the split part.

```python
df.groupby('Department')['salary'].mean()
```

Split → Apply → Combine

groupby() + aggregation

```python
df.groupby('Department')['salary'].sum()
```

```python
df.groupby('Department')['salary'].min()
```

Important difference: count() vs size()
count() counts non-missing values.
size() counts rows, including rows containing NaN.

```python
df.groupby('Department')['salary'].count()
```

```python
df.groupby('Department')['salary'].size()
```

Multiple aggregations

```python
df.groupby('Department')['salary'].agg(['mean', 'max', 'min', 'sum'])
```

```python
ipl.groupby('Season')['Margin'].agg(['mean', 'max', 'min', 'sum'])
```

Different functions for different columns

```python
df = pd.DataFrame({
    'Department': ['IT', 'IT', 'HR', 'HR'],
    'Salary': [50000, 60000, 40000, 45000],
    'Age': [25, 30, 28, 35]
})
df
```

```python
df.groupby('Department').agg({
    'Salary': 'mean',
    'Age': 'max'
})
```

```python
diabetes
```

```python
diabetes.groupby('Pregnancies').agg({
   'Glucose':'mean',
   'BloodPressure' :'min' ,
   'SkinThickness':'max'
})
```

```python
diabetes.groupby('Pregnancies').agg({
   'Glucose':'mean',
   'BloodPressure' :['min','max'] ,
   'SkinThickness':'max'
})
```

Named aggregation

```python
df.groupby('Department').agg(
    average_salary=('Salary', 'mean'),
    maximum_age=('Age', 'max')
)
```

```python
ipl.groupby('Season').agg(
     mean_margin=('Margin','mean')
)
```

Grouping by multiple columns

```python
df = pd.DataFrame({
    'Department': ['IT', 'IT', 'IT', 'HR', 'HR'],
    'Gender': ['M', 'F', 'M', 'F', 'M'],
    'Salary': [50000, 60000, 55000, 40000, 45000]
})
df
```

```python
df.groupby(['Department', 'Gender'])['Salary'].mean()
```

as_index

```python
df.groupby('Department')['Salary'].mean()
```

```python
df.groupby('Department', as_index=False)['Salary'].mean()
```

sort

```python
df.groupby('Department', sort=False)['Salary'].mean()
```

```python
df.groupby('Department',sort=False)['Salary'].mean().sort_values()
```

```python
type(df.groupby('Department')[['Salary']].mean())
```

```python
df.groupby('Department')['Salary'].idxmax()
```

```python
df.loc[df.groupby('Department')['Salary'].idxmax()]
```

transform() with groupby()

```python
df['dept_mean'] = (
    df.groupby('Department')['Salary'].transform('mean')
)
df
```

filter()
filter() allows you to keep or remove entire groups based on a condition.

```python
df.groupby('Department').filter(
    lambda x: len(x) >= 2
)
```

apply()

```python
def highest_salary(group):
    return group.loc[group['Salary'].idxmax()]

df.groupby('Department').apply(highest_salary)
```

Iterating over groups

```python
for name, group in df.groupby('Department'):
    print(name)
    print(group)
```

get_group()

```python
grouped = df.groupby('Department')
grouped.get_group('IT')
```

Missing values and dropna

```python
df.groupby('Department', dropna=False).size()
```

```python
df.groupby('Department').size()
```
