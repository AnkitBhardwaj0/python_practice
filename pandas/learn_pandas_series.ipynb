### Pandas Series

A Pandas Series is like a column in a table. It is a 1-D array holding data of any type.

### Importing Pandas

```python
import pandas as pd
import numpy as np
```

### Series from lists

```python
#string
country=['India','China','Pakistan','USA','Nepal','Srilanka']
pd.Series(country)
```

```python
#integers
runs=[31,42,65,88,100]
pd.Series(runs)
```

```python
#custom index
from operator import index


marks=[67,38,99,83,100]
subjects=['english','science','maths','social_science','hindi']
pd.Series(marks , index=subjects)
```

```python
#setting a name
marks=pd.Series(marks,index=subjects,name='Ankit marks')
marks
```

### Series from dict
