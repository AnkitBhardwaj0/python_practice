```python
# problem in vectorized opertions in vanilla python
s = ['cat','mat',None,'rat']
[i.startswith('c') for i in s]
```

```python
# How pandas solves this issue?
s = pd.Series(['cat','mat',None,'rat'])
# string accessor
s.str.startswith('c')

# fast and optimized
```

```python
print(s.dtype)
```

```python
print(s.str.startswith('c').iloc[2])
```

```python
print(type(s.iloc[2]))
```

```python
print(s.iloc[2] is pd.NA)
```

```python
s1 = pd.Series(['cat', 'mat', None, 'rat'], dtype='str')
s2 = pd.Series(['cat', 'mat', None, 'rat'], dtype='string')

print(s1)
print(s1.dtype)
print(type(s1.iloc[2]))

print()

print(s2)
print(s2.dtype)
print(type(s2.iloc[2]))
```

```python
# split -> get
df['Name'].str.split(',')
```

```python
df['lastname']=df['Name'].str.split(',').str.get(0)
df.head()
```

```python
df['Name'].str.split(',').str.get(1)
```

```python
df['Name'].str.split(',').str.get(1).str.split(' ')
```

```python
df['Name'].str.split(',').str.get(1).str.strip().str.split(' ',n=1)#n no of split if n=1 then only one split from starting
```

```python
df['Name'].str.split(',').str.get(1).str.strip().str.split(' ',n=1,expand=True)#convert into dataframe
```

```python
df[['title','firstname']]=df['Name'].str.split(',').str.get(1).str.strip().str.split(' ',n=1,expand=True)
```

```python
df.head()
```

```python
df['title'].value_counts()
```

```python
# replace
df['title']=df['title'].str.replace('Ms.','Miss.')
df['title']=df['title'].str.replace('Mlle.','Miss.')
```

```python
df['title'].value_counts()
```

```python
# filtering
# startswith/endswith
df[df['firstname'].str.startswith('A')]
```

```python
df[df['firstname'].str.endswith('A')]
```

```python
# isdigit/isalpha...
df[df['firstname'].str.isdigit()]
```

```python
# applying regex
# contains
# search john -> both case
df[df['firstname'].str.contains('jhon',case=False)]
```

```python
# find lastnames with start and end char vowel
df[df['lastname'].str.contains('^[aeiouAEIOU].+[aeiouAEIOU]$')]
```

```python
# find lastnames with start and end char consonant
df[df['lastname'].str.contains('^[^aeiouAEIOU].+[^aeiouAEIOU]$')]#regular expression
```
