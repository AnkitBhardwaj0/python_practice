itertools

itertools is a built-in Python module that provides functions for working with iterators.

count()

Creates an infinite sequence of numbers.

```python
import itertools
```

```python
x = itertools.count(1)
print(next(x))
print(next(x))
print(next(x))
print(next(x))
```

```python
x = itertools.count(10, 2) #2 step 
print(next(x))
print(next(x))
print(next(x))
```

cycle()

Repeats an iterable again and again.

```python
x = itertools.cycle([1, 2, 3])
print(next(x))
print(next(x))
print(next(x))
print(next(x))
print(next(x))
```

repeat()  :- Repeats the same value.

```python
x = itertools.repeat("Python", 3)
print(list(x))
```

chain() :- 

iterate through both as if they were one sequence.

```python
a = [1, 2, 3]
b = [4, 5, 6]
result = itertools.chain(a, b)
print(list(result))
```

combinations() :- a pair order doesn't matter

```python
students = ['A', 'B', 'C']
x = itertools.combinations(students, 2)
print(list(x))
```

permutations()

Similar to combinations, but order matters.

```python
students = ['A', 'B', 'C']
x = itertools.permutations(students, 2)
print(list(x))
```

product()

Creates a Cartesian product.

```python
colors = ['Red', 'Blue']
sizes = ['S', 'M']
x = itertools.product(colors, sizes)
print(list(x))
```

groupby() :-

Groups consecutive elements having the same key.

```python
data = [
    ('A', 10),
    ('A', 20),
    ('B', 30),
    ('B', 40)
]
for key, group in itertools.groupby(data, key=lambda x: x[0]):
    print(key, list(group))
```

islice()

work like slicing ,but for iterators

```python
x = itertools.count(1)
print(list(itertools.islice(x, 5)))
```

```python
x = itertools.count(1)
print(list(itertools.islice(x, 2, 8)))
```

itertools

count() → infinite numbers

cycle() → repeat sequence

repeat() → repeat one value

chain() → combine iterables

combinations() → selection, order doesn't matter

permutations() → arrangement, order matters
product() → every possible combination

groupby() → group consecutive elements

islice() → slice an iterator
